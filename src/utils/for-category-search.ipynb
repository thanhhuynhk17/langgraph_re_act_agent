```python
def type_values_in_type_of_food(df: pd.DataFrame) -> list[list]:
    """_summary_

    Args:
        df (pd.DataFrame): _description_

    Returns:
        list[list]: _description_
    """
    vals_counts = df["type_of_food"].value_counts()
    return np.column_stack(([name.lower() for name in vals_counts.index.tolist()], vals_counts.values.tolist())).tolist()

np.array(type_values_in_type_of_food(df))[::,0]
```

```python
import pandas as pd

def search_type_category_and_people(df: pd.DataFrame, value_type_of_food: str, value_option: str, num_people: int):
    
    # value_type_of_food in ('món cá', 'món khai vị', 'món ăn chơi', 'món rau', 'món gỏi',
    #    'món gà, vịt & trứng', 'món tôm & mực', 'món xào',
    #    'nước mát nhà làm', 'lẩu', 'món thịt', 'món sườn & đậu hũ',
    #    'món canh', 'các loại khô', 'tráng miệng')
    
    # value_option là yêu cầu khách hàng
    
    # num_people là số lượng người ăn mà khách yêu cầu nếu không có thể thay đổi (1,2,3,4,5,6,7)
    
    value_option = value_option.lower()
    # 🔹 1. Hold fixed type_of_food filter
    mask_type = df["type_of_food"].str.lower().str.contains(value_type_of_food.lower(), na=False)

    # 🔹 2. Search value_option across other columns (excluding type_of_food itself)
    search_cols = df.columns.drop(["type_of_food", "number_of_people_eating"])

    mask_text = df[search_cols].apply(
        lambda col: col.astype(str).str.lower().str.contains(value_option.lower(), na=False)
    )

    mask_any = mask_text.any(axis=1)
    
    mask_people = df["number_of_people_eating"].str.contains(str(num_people), na=False)

    # 🔹 3. Combine filters
    result = df[mask_type & mask_any & mask_people].reset_index(drop=True)

    return result

df_fil = search_type_category_and_people(df, "MÓN CÁ", "mặn", 2)
df_fil
```
